```python
# 14 moving left
for i in range(0, len(array_2)):
    next, array_2[-i] = array_2[-i], next
array_2
```

```python
# 14 moving left again
print(array_2)
for i in range(0, len(array_2)-1):
    next = array_2[i]
    array_2[i] = array_2[i+1]
    array_2[i+1] = next
array_2
```

```python
# 14 moving right
print(array_2)
last = array_2[-1]
for i in range(len(array_2) - 1, 0, -1):
    array_2[i] = array_2[i-1]
array_2[0] = last
array_2
```
